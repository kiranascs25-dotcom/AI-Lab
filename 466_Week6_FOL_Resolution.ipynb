```python
clauses = [
    "NOT Food(x) OR Likes(John,x)",
    "NOT Eats(y,z) OR Killed(y) OR Food(z)",
    "Eats(Anil,Peanuts)",
    "Alive(Anil)",
    "NOT Alive(x) OR NOT Killed(x)",
    "NOT Likes(John,Peanuts)"
]

print("Knowledge Base:")
for i, clause in enumerate(clauses, 1):
    print(f"C{i}: {clause}")

print("\nResolution Steps:")

r1 = "NOT Food(Peanuts)"
print("R1:", r1)

r2 = "NOT Eats(Anil,Peanuts) OR Killed(Anil)"
print("R2:", r2)

r3 = "Killed(Anil)"
print("R3:", r3)

r4 = "NOT Killed(Anil)"
print("R4:", r4)

r5 = "{}"
print("R5:", r5)

print("\nEmpty clause obtained!")
print("Contradiction found.")
print("Hence, John likes peanuts is PROVED.")
```
