```python
import random
import math

def calculate_cost(state):
    conflicts = 0

    for i in range(8):
        for j in range(i + 1, 8):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def get_neighbor(state):
    neighbor = state[:]
    col = random.randint(0, 7)
    neighbor[col] = random.choice(
        [r for r in range(8) if r != neighbor[col]]
    )
    return neighbor

def simulated_annealing():
    random.seed(42)

    for restart in range(100):

        current = [random.randint(0, 7) for _ in range(8)]
        T = 100.0

        while T > 0.01:

            current_cost = calculate_cost(current)

            if current_cost == 0:
                return current

            next_state = get_neighbor(current)
            next_cost = calculate_cost(next_state)

            delta_E = current_cost - next_cost

            if delta_E > 0:
                current = next_state

            else:
                probability = math.exp(delta_E / T)

                if random.random() < probability:
                    current = next_state

            T = T * 0.995

    return None

solution = simulated_annealing()

print("SIMULATED ANNEALING - 8 QUEENS")
print("--------------------------------")

if solution:
    print("Solution Found!")
    print("Queen Positions:", [x + 1 for x in solution])
    print("Cost:", calculate_cost(solution))

    print("\nChess Board:")

    for row in range(7, -1, -1):
        for col in range(8):
            if solution[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()

else:
    print("Solution not found")
```
