```python
def alpha_beta(node, alpha, beta, maximizing):
    global pruned, visited

    if not isinstance(node, list):
        visited.append(node)
        return node

    if maximizing:
        best = float('-inf')

        for i, child in enumerate(node):
            value = alpha_beta(child, alpha, beta, False)
            best = max(best, value)
            alpha = max(alpha, best)

            if alpha >= beta:
                for remaining in node[i + 1:]:
                    get_pruned(remaining)
                break

        return best

    else:
        best = float('inf')

        for i, child in enumerate(node):
            value = alpha_beta(child, alpha, beta, True)
            best = min(best, value)
            beta = min(beta, best)

            if beta <= alpha:
                for remaining in node[i + 1:]:
                    get_pruned(remaining)
                break

        return best


def get_pruned(node):
    if isinstance(node, list):
        for child in node:
            get_pruned(child)
    else:
        pruned.append(node)

tree = [
    [[10, 9], [14, 18]],
    [[5, 4], [50, 3]]
]

pruned = []
visited = []

result = alpha_beta(
    tree,
    float('-inf'),
    float('inf'),
    True
)

print("ALPHA-BETA PRUNING")
print("------------------------")
print("Visited Nodes:", visited)
print("Pruned Nodes:", pruned)
print("Root Node Value:", result)
print("Optimal Path: Root -> Left -> Left -> 10")
```
